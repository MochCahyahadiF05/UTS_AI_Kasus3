**Menentukan Kasus Berdasarkan NIM**

Nomor kasus ditentukan dengan menjumlahkan seluruh digit NIM, kemudian menghitung sisa pembagiannya dengan 10. Pada kode berikut, NIM disimpan sebagai teks agar setiap digit dapat dibaca satu per satu. Setiap digit diubah menjadi bilangan menggunakan int(), lalu dijumlahkan menggunakan sum().

```python
nim = "23552011347"

jumlah_digit = sum(int(digit) for digit in nim)
sisa = jumlah_digit % 10

if sisa == 0:
    kasus = 10
else:
    kasus = sisa

print("NIM          =", nim)
print("Jumlah digit =", jumlah_digit)
print("Nomor kasus  =", kasus)
```

**Memuat Dataset Penguin**

Berdasarkan hasil perhitungan NIM, dataset yang digunakan adalah kasus03_penguins.csv. Kode berikut membuat folder dataset, kemudian mendownload file CSV sesuai kasus yang di tentukan,kemudian membacanya menggunakan Pandas. Lima baris pertama ditampilkan untuk mengenali isi data. Setiap baris mewakili satu penguin, sedangkan kolom species menunjukkan spesies yang akan diprediksi.

```python
from pathlib import Path
from urllib.request import urlretrieve
import pandas as pd

Path("dataset").mkdir(exist_ok=True)

url = (
    "https://raw.githubusercontent.com/Muhammad-Ikhwan-Fathulloh/"
    "Artificial-Intelegent-Course-Bank/main/UTS/dataset/kasus03_penguins.csv"
)

file_dataset = "dataset/kasus03_penguins.csv"
urlretrieve(url, file_dataset)

df = pd.read_csv(file_dataset)

print("Jumlah baris :", df.shape[0])
print("Jumlah kolom :", df.shape[1])
display(df.head())
```

**Audit Kualitas Data**

Sebelum melatih model, dataset diperiksa untuk mengetahui tipe data, jumlah nilai kosong, jumlah baris duplikat, dan kemungkinan ukuran fisik yang tidak wajar. Panjang paruh, kedalaman paruh, panjang sirip, dan massa tubuh seharusnya berupa bilangan positif. Jumlah penguin pada setiap spesies juga diperiksa untuk mengetahui apakah ketiga kelas memiliki jumlah anggota yang seimbang.

```python
import numpy as np

kolom_ukuran = [
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
]

print("Tipe data dan jumlah nilai kosong")
display(pd.DataFrame({
    "Tipe data": df.dtypes,
    "Jumlah kosong": df.isna().sum()
}))

print("Jumlah baris duplikat:", df.duplicated().sum())

angka = df[kolom_ukuran].apply(pd.to_numeric, errors="coerce")
tidak_valid = df[kolom_ukuran].notna() & (
    angka.isna() | (angka <= 0) | ~np.isfinite(angka)
)

print("\nPemeriksaan ukuran fisik")
display(pd.DataFrame({
    "Nilai minimum": angka.min(),
    "Nilai maksimum": angka.max(),
    "Jumlah tidak valid": tidak_valid.sum()
}))

print("Jumlah penguin per spesies")
display(df["species"].value_counts())

print("Tahun pengamatan:", sorted(df["year"].unique()))
```
